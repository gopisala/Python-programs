```python
#Recursive Function_Factorial
def factorial(n):
  if n==0 or n==1:
    return 1
  return n*factorial(n-1)
print("Factorial of 6:",factorial(6))

```

```python
#Function with type hints
def calculate_bmi(weight:float,height:float) ->str:
  bmi=weight/(height**2)
  if bmi<18.5:
    return f"BMI:{bmi:.1f}-Underweight"
  elif bmi<25:
    return f"BMI:{bmi:.1f}-Normal"
  else:
    return f"BMI:{bmi:.1f}-Overweight"
print(calculate_bmi(70,1.75))
```

```python
#Function chaining
def remove_spaces(text):
  return text.strip()
def to_uppercase(text):
  return text.upper()
def add_exclamation(text):
  return text+"!"
def process(text):
  return add_exclamation(to_uppercase(remove_spaces(text)))
print(process("Hello world"))
```

```python
#Fibonacci using functions
def fibonacci(n):
  sequence=[]
  a,b=0,1
  for_in range(n):
    sequence.append(a)
    a,b=b,a+b
   return sequence
def print_fibonacci(n):
  seq=fibonacci(n)
  print("Fibonacci:",seq)
print_fibonacci(10)
```

```python
#HIGHER-ORDER FUNCTIONs_map,filter,reduce
from functools import reduce
numbers=[1,2,3,4,5,6]
squared=list((map(lambda x:x**2,numbers)))
evens=list(filter(lambda x:x%2==0,numbers))
product=reduce(lambda x,y:x*y,numbers)
print("Squared:",squared)
print("Evens:",evens)
print("Total:",product)
```

```python
# Decorator function
def timer_decorator(func):
  import time
  def wrapper(*args,**kwarges):
    start=time.time()
    result=func(*args,**kwargs)
    end=time.time()
    print(f"{func._name_}ran in {end-start:,4f}s")
    return result
  return wrapper
@timer_decorator
def compute_sum(n):
  return sum(range(n))
  print(compute_sum(1000000))
```

```python
#NESTED FUNCTIONS
def bank_account(balance):
  def deposit(amount):
    return balance+amount
  def withdraw(amount):
    if amount>balance:
      return "Insufficient funds"
    return balance-amount
print("Deposit 500:",deposit(500))
print("Withdraw 200:",withdraw(200))
print("Withdraw 999",withdraw(999))
bank_account(300)
```

```python
#Function with Exception handling
def safe_divide(a,b):
  try:
    return a/b
  except ZeroDivisionError:
    return "Error:Division by zero"
  except TypeError:
      return "Error:Invalid types"
print(safe_divide(10,2))
print(safe_divide(10,0))
print(safe_divide(10,"x"))
```

```python
#Generator function
def prime_genetor(limit):
  for num in range(2,limit+1):
    is_prime=all(num%i!=0 for i in range(2,int(num**0.5)+1))
    if is_prime:
      yield num
primes=list(prime_genetor(30))
print("Primes up to 30:",primes)
```

```python
#FUNCTION PASSING AS ARGUMENT
def apply_operation(a,b,operation):
  return operation(a,b)
def add(x,y):    return x+y
def subtract(x,y): return x-y
def multiply(x,y):return x*y
for op in [add,subtract,multiply]:
  print(f"{op._name_}(10,3)={apply_operation(10,3,op)}")
```

```python
#Memoization using functions cache
from functools import iru_cache

@iru_cache(maxsize=None)
def fib(n):
  if n<=1:
    return n
  return fib(n-1)+fib(n-2)
for i in [10,20,30,40]:
  print(f"fib{i}={fib(i)}")

```

```python
#Parital function application
from functools import partial
def power(base,exponent):
  return base**exponent
square=partial(power,exponent=2)
cube=partial(power,exponent=3)

for n in [2,3,4,5]:
  print(f"{n}^2={square(n)} {n}^3={cube(n)}")
```
