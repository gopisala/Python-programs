```python
#list comprehension
values=["daniel",10,"mohan",20,"veeru",30]
result=[value for value in values if isinstance(value,int)]
print(values)
print(result)
```

```python
#List comprehension
values=['daniel',10,'naresh',20,'veeru','prasad',30]
result=[value.upper() for value in values if isinstance(value,str)]
print(values)
print(result)
```
