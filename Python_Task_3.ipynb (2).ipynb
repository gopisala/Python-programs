```python
#write a program to take a string input and check if it is a palindrome.
text=input("Enter a string:").strip().lower()
if text==text[::-1]:
  print(f"{text}' is a Palindrome")
else:
    print(f"{text}' is not a palindrome")
```

```python
#write a program to use operator precedence:evaluate(2+3)*4**2/8-1 and show step ny step results
#full expression
result=(2+3)*4**2/8-1
print("Full expression result:",result)
#set by step
step1=2+3    #5 (parentheses first)
step2=4**2     #16  (exponentiation)
step3=step1*step2   #80(multiplication)
step4=step3/8    #10(divion)
step5=step4-1   #9(subtraction)
print(f"Step 1:(2+3={step1})")
print(f"step 2: 4**2={step2}")
print(f"step 3:step1*step2={step3}")
print(f"step 4:step3/8={step4}")
print(f"step 5:step4-1={step5}")
```
